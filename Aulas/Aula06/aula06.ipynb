```python
def is_number(value):
    try:
        float(value) or int(value)
        return True
    except ValueError:
        return False
```

```python
def quadrado(x = float) -> float:
    if not is_number(x):
        raise ValueError("O valor deve ser um número.")
    return x * x

def cubo(x = float) -> float:
    if isinstance(x, str):
        raise ValueError("O valor deve ser um número.")
    return quadrado(x) * x
```

```python
def parte_inteira(dividendo, divisor=1):
    if divisor == 0:
        raise ValueError("O divisor não pode ser zero.")
    return dividendo // divisor
```

```python
print(parte_inteira(18))
```

```python
def sao_iguais(x, y):
    if x == y:
        return print("Os valores são iguais.")
    else:
        return print("Os valores são diferentes.")
```

```python
sao_iguais(5, 5.01)
sao_iguais(5,11)
```
