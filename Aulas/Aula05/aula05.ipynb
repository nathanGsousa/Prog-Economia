```python
def chamar_a_mae():
    """Função chamar a mãe para ajudar com o Python.
    
    """
    a = 0
    while a < 10:
        a += 1
        print("Mãe, me ajuda com o Python!!!!")
```

```python
chamar_a_mae()
```

```python
def dividir(a, b=1):
    """Função para dividir dois números.
    
    Args:
        a: O número a ser dividido.
        b: O número pelo qual dividir. Padrão é 1.
    
    Returns:
        O resultado da divisão de a por b.
    """
    return a / b
```

```python
class calculadora:
    """Classe para realizar operações matemáticas básicas."""
    
    def somar(self, a, b):
        """Método para somar dois números."""
        return a + b
    
    def subtrair(self, a, b):
        """Método para subtrair dois números."""
        return a - b
    
    def multiplicar(self, a, b=1):
        """Método para multiplicar dois números."""
        return a * b
    
    def dividir(self, a, b=1):
        """Método para dividir dois números."""
        if b == 0:
            raise ValueError("Não é possível dividir por zero.")
        return a / b
```

```python
a = calculadora()

a.somar(10, 5)
```

## Estrutura Condicional

#### Decisão IF, ELIF, ELSE

```python
def voce_tem_mae():
    """Função para verificar se a pessoa tem mãe."""
    
    resposta = input("Você tem mãe? (sim/não) ")
    
    if resposta.lower() == "sim":
        chamar_a_mae()
    else:
        print("Sinto muito por você. Espero que encontre alguém para te ajudar com o Python.")


voce_tem_mae()
```

```python
lista = [1, 2, 3, 4, 5]

def verificar_numero_na_lista(numero):
    """Função para verificar se um número está na lista."""
    
    if numero in lista:
        print(f"O número {numero} está na lista.")
    else:
        print(f"O número {numero} não está na lista.")


verificar_numero_na_lista(3)
verificar_numero_na_lista(6)
verificar_numero_na_lista(10)
```
