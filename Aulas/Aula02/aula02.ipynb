## Python operadores aritmieticos

```python
x = 3

def greet(name):
    return f"Hello, {name}!"

for i in range(x):
    print(greet("Nathan"))
```

```python
while x > 0:
    print(f"x is currently: {x}")
    x -= 1

def factorial(n):
    if n == 0:
        return 1
    else:
        return n * factorial(n - 1)

print(factorial(5))
```
